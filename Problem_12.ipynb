<div style = "font-family:'Times New Roman';font-size:20px">
<p><i>01.Find all elements greater than 10.
<p>02.Find all even numbers.</p>
<p>03.Replace every number less than 10 with 0.</p>
<p>04.Calculate the average of the elements greater than 10.</i></p>
</div>

```python
import numpy as np
array = np.array([2,30,45,4,678,87,38,7,90,5,789])

Greater_than_10 = array[array > 10]
# 01.
print("Numbers which are greater than 10:",Greater_than_10)
# 02.
Even_num = array[array % 2 == 0]
print("All the even numbers:",Even_num)
# 03. 
array[array < 10] = 0
print("Replace the elements which are less than 10:",array)
# 04.
average = np.average(array[array > 10])
print("Average of the elements greater than 10:",average)
```
