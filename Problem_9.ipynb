#### Given:
#### python arr = np.array([10, 20, 30, 40, 50])
#### Find the *maximum* and *minimum* values

```python
import  numpy as np
arr = np.array([10,20,30,40,50])
maximum = max(arr)
minimum = min(arr)
print("Maximum element in the given array is :",maximum)
print("Minimum element in the given array is :",minimum)
```
